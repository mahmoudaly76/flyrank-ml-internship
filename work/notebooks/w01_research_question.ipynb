## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load the dataset directly from the raw GitHub link to avoid folder path errors
url = "https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

# Calculate required numbers
total_pages = df.shape[0]
declining_pages = df[df["trend_direction"].str.lower() == "down"].shape[0]
decay_rate = (declining_pages / total_pages) * 100

print(f"Data Evidence for this Lane:")
print(f"1. Total pages analyzed: {total_pages:,}")
print(f"2. Pages currently declining: {declining_pages:,}")
print(f"3. Overall decay rate: {decay_rate:.1f}%")
print("Conclusion: With more than half the pages losing traffic, a predictive model is highly justified.")
```
